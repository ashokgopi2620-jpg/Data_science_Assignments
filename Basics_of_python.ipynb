Prime Number Checker
This program takes a user-input number and checks if it's prime by testing divisors from 2 to the square root of the number.

```python
import math

def is_prime(n):
    if n <= 1:
        return False
    if n <= 3:
        return True
    if n % 2 == 0 or n % 3 == 0:
        return False
    i = 5
    while i * i <= n:
        if n % i == 0 or n % (i + 2) == 0:
            return False
        i += 6
    return True

num = int(input("Enter a number: "))
if is_prime(num):
    print(f"{num} is prime.")
else:
    print(f"{num} is not prime.")
```

Product Quiz Game
Generates two random numbers (1-10), asks for their product, and checks the answer.

```python
import random

num1 = random.randint(1, 10)
num2 = random.randint(1, 10)
product = num1 * num2

user_answer = int(input(f"What is {num1} * {num2}? "))
if user_answer == product:
    print("Correct!")
else:
    print(f"Wrong. The product is {product}.")
```

Hospital Billing System
Calculates bill for given inputs: 4 days stay, 3 tests, 2400 medicine; applies 10% discount if over 10,000.

```python
days = 4
room_per_day = 2000
doctor_fee = 1500
tests = 3
test_cost = 300
medicine = 2400

total = (days * room_per_day) + doctor_fee + (tests * test_cost) + medicine
print(f"Total before discount: {total}")

if total > 10000:
    discount = total * 0.10
    total -= discount
    print(f"Discount applied: {discount}")
    print(f"Final bill: {total}")
else:
    print(f"Final bill: {total}")
```

```python
import math
import random

def prime_checker():
    num = int(input("Enter a number: "))
    if num <= 1: return "Not prime"
    for i in range(2, int(math.sqrt(num)) + 1):
        if num % i == 0: return "Not prime"
    return "Prime"

def product_quiz():
    n1, n2 = random.randint(1,10), random.randint(1,10)
    ans = int(input(f"{n1} * {n2} = ? "))
    return "Correct!" if ans == n1*n2 else f"Wrong, it's {n1*n2}"

def squares_even():
    print("Squares of even numbers 100-200:")
    for num in range(100, 201, 2):
        print(num**2)

def hospital_bill():
    days = int(input("Days: ")) or 4
    tests = int(input("Tests: ")) or 3
    med = float(input("Medicine: ")) or 2400
    total = (days*2000) + 1500 + (tests*300) + med
    if total > 10000: total *= 0.9
    return f"Final bill: ₹{total:.2f}"

while True:
    print("\n1. Prime  2. Quiz  3. Squares  4. Bill  5. Exit")
    choice = input("Choose: ")
    if choice == '1': print(prime_checker())
    elif choice == '2': print(product_quiz())
    elif choice == '3': squares_even()
    elif choice == '4': print(hospital_bill())
    elif choice == '5': break
```
